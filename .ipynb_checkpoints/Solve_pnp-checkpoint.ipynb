```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
rgb_img = cv2.imread('data/rgb/rgb_50.png')
gray = cv2.cvtColor(src=rgb_img, code=cv2.COLOR_BGR2GRAY)


# rgb_img = cv2.cvtColor(rgb_img, cv2.COLOR_BGR2RGB)
# gray = cv2.cvtColor(src=rgb_img, code=cv2.COLOR_BGR2GRAY)
plt.imshow(rgb_img)
plt.show()
```

```python
img_hsv = cv2.cvtColor(rgb_img,cv2.COLOR_RGB2HSV)
H,S,V = cv2.split(img_hsv)
hist = cv2.calcHist(H,[2],None,[256],[0,256])
# H、S、V范围一：

lower1 = np.array([0,43,46])
upper1 = np.array([10,255,255])
mask1 = cv2.inRange(img_hsv, lower1, upper1)       # mask1 为二值图像

res1 = cv2.bitwise_and(rgb_img, rgb_img, mask=mask1)

    # H、S、V范围二：
lower2 = np.array([156,43,46])
upper2 = np.array([180,255,255])
mask2 = cv2.inRange(img_hsv, lower2, upper2)
res2 = cv2.bitwise_and(rgb_img,rgb_img, mask=mask2)
mask3 = mask1 + mask2
                  
plt.imshow(mask3)
```

```python
edges = cv2.Canny(image=mask1, threshold1=500, threshold2=1500, apertureSize=3)
plt.imshow(edges)

kernel = np.ones((4,5),dtype = np.uint8)
opening = cv2.morphologyEx(mask3, cv2.MORPH_OPEN, kernel)
opening = cv2.morphologyEx(opening, cv2.MORPH_OPEN, kernel)
opening = cv2.morphologyEx(opening, cv2.MORPH_OPEN, kernel)
opening = cv2.morphologyEx(opening, cv2.MORPH_OPEN, kernel)

plt.imshow(opening)
```

```python
import copy
rgb_img2 = copy.deepcopy(rgb_img)
```

```python
lines = cv2.HoughLinesP(image=opening, rho=1, theta=np.pi/180, threshold=50, minLineLength=100, maxLineGap=2)
for line in lines:
    
    for x1,y1,x2,y2 in line:
        cv2.line(rgb_img2,(x1,y1),(x2,y2),(0,200,0),2)
```

```python
# 初始化SIFT检测器
import cv2
import numpy as np
import matplotlib.pyplot as plt

sift = cv2.SIFT_create()
img1 = cv2.imread('img/rgb_1.png')
img2 = cv2.imread('img/ir_1.jpg')

# 寻找关键点和描述符
kp1, des1 = sift.detectAndCompute(img1, None)
kp2, des2 = sift.detectAndCompute(img2, None)

# 初始化FLANN匹配器
FLANN_INDEX_KDTREE = 0
index_params = dict(algorithm=FLANN_INDEX_KDTREE, trees=5)
search_params = dict(checks=100)
flann = cv2.FlannBasedMatcher(index_params, search_params)

# 使用KNN算法进行匹配
matches = flann.knnMatch(des1, des2, k=4)

# 进行比率测试，筛选出较好的匹配点对
good_matches = []
for m, n in matches:
    if m.distance < 0.5 * n.distance:
        good_matches.append(m)

# 获取匹配点的坐标
src_pts = np.float32([kp1[m.queryIdx].pt for m in good_matches]).reshape(-1, 1, 2)
dst_pts = np.float32([kp2[m.trainIdx].pt for m in good_matches]).reshape(-1, 1, 2)

# 计算仿射变换矩阵
M, _ = cv2.estimateAffinePartial2D(src_pts, dst_pts)

# 打印输出仿射变换矩阵
print("仿射变换矩阵:")
print(M)

# 绘制匹配点
matches_img = cv2.drawMatches(img1, kp1, img2, kp2, good_matches, None, flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS)

# 显示匹配点图像
cv2.imshow("Matches", matches_img)
cv2.waitKey(0)
```

```python
inputImage = cv2.imread('data/rgb/rgb_02.jpg')
# inputImage = cv2.cvtColor(src=rgb_img, code=cv2.COLOR_BGR2GRAY)


# rgb_img = cv2.cvtColor(rgb_img, cv2.COLOR_BGR2RGB)
# gray = cv2.cvtColor(src=rgb_img, code=cv2.COLOR_BGR2GRAY)
plt.imshow(inputImage)
plt.show()
```

```python
# 创建一个 qrCodeDetector 对象
qrDecoder = cv2.QRCodeDetector()

data, bbox, rectifiedImage = qrDecoder.detectAndDecode(inputImage)

if len(data) > 0:
    print("Decoded Data : {}".format(data))
    display(inputImage, bbox)
    rectifiedImage = np.uint8(rectifiedImage);
    cv2.imshow("Rectified QRCode", rectifiedImage);
else:
    print("QR Code not detected")
    cv2.imshow("Results", inputImage)
cv2.imwrite("output.jpg", inputImage)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import time
start = time.time()
# Detect and decode the QRs within the image
QRs = qreader.detect(image=inputImage)
end = time.time()

# Print the results
print('detection time: ', end-start)

cxcy = QRs['cxcy']
```

```python
from qrdet import QRDetector
import cv2

detector = QRDetector(model_size='s')
image = inputImage

start = time.time()
detections = detector.detect(image=image, is_bgr=True)
end = time.time()
print('detection time: ', end-start)

# Draw the detections
for detection in detections:
    x1, y1, x2, y2 = detection['bbox_xyxy'].astype(np.int16)
    confidence = detection['confidence']
    cv2.rectangle(image, (x1, y1), (x2, y2), color=(0, 255, 0), thickness=2)
    cv2.putText(image, f'{confidence:.2f}', (x1, y1 - 10), fontFace=cv2.FONT_HERSHEY_SIMPLEX,
                fontScale=1, color=(0, 255, 0), thickness=2)
plt.imshow(image)
```
